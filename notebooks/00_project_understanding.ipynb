## 1. Business Problem

Customer churn happens when an existing customer stops using a company's product or service.

Businesses need to know:

- Which customers are likely to churn?
- Which customers are at highest risk?
- Why is a customer likely to churn?
- Which customers should be prioritized?
- What retention action should be taken?

A simple `Churn / No Churn` prediction is not enough for business decision-making.

## 2. Project Objective

RetainIQ will aim to provide:

**Prediction + Explanation + Prioritization + Retention Recommendation + Business Insight**

Main capabilities:

- Predict customer churn probability
- Classify customers by risk level
- Explain why a customer is at risk
- Identify important churn factors
- Recommend retention actions
- Provide customer-level insights
- Provide business analytics
- Support individual and batch predictions
- Store customer and prediction information
- Present results through an interactive dashboard

## 3. Machine Learning Problem

### Learning Type
**Supervised Learning**

The model learns from historical customer data where the churn outcome is already known.

### Problem Type
**Binary Classification**

Target:

- `0` → No Churn
- `1` → Churn

### Important distinction

- **Churn Prediction** → predicted class: Churn / No Churn
- **Churn Probability** → estimated likelihood of churn, e.g. 0.87
- **Risk Level** → business category derived from probability

## 5. Planned Models

Models will be investigated progressively:

1. Logistic Regression — interpretable baseline
2. Decision Tree
3. Random Forest
4. XGBoost

### Important Project Principle

The final model will **not** be selected only because it has the highest accuracy.

Model selection will depend on:

- Classification performance
- Churn recall
- Precision / Recall trade-off
- F1 Score
- ROC-AUC
- PR-AUC
- Business objective

## 6. Why Accuracy Alone Is Not Enough

Churn datasets may be imbalanced.

Example:

```text
10,000 customers

9,500 → No Churn
500   → Churn
```

A useless model that predicts every customer as `No Churn` would still achieve:

```text
Accuracy = 95%
```

but detect **zero actual churners**.

Therefore RetainIQ will later evaluate:

- Confusion Matrix
- Precision
- Recall
- F1 Score
- ROC-AUC
- PR-AUC

## 7. Explainable AI

RetainIQ will use **SHAP** to explain model predictions.

Instead of only showing:

```text
Churn Probability = 87%
```

the platform should also explain possible contributing factors such as:

```text
↑ Low satisfaction
↑ Long inactivity
↑ Complaint history
↑ Low engagement
↓ Longer tenure
```

The exact explanations must come from the trained model and actual data.

## 8. Customer Risk Segmentation

Initial planned probability bands:

```text
0–25%     LOW
25–50%    MEDIUM
50–75%    HIGH
75–100%   CRITICAL
```

These are **initial business rules**, not permanent values.

They may be adjusted later after studying:

- Model probabilities
- Threshold selection
- Business trade-offs

## 9. Retention Recommendation Engine

The system will convert risk information into possible actions.

Example:

```text
Risk: CRITICAL

Main Risk Factors:
- Low satisfaction
- Long inactivity
- Complaint history

Suggested Actions:
- Resolve complaint
- Re-engagement campaign
- Personalized retention offer
```

The initial recommendation engine can use transparent business rules.
